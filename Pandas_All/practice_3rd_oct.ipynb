```python
import pandas as pd
import plotly.express as px
```

```python
dtc_df=df=pd.read_csv(r"F:\KPIT\Pandas_All\Datasets\dtc_codes.csv",index_col="timestamp")
sensor_df=pd.read_csv(r"F:\KPIT\Pandas_All\Datasets\ecu_sensor_data.csv",index_col="timestamp")
display(dtc_df.sample(5))
display(sensor_df.sample(5))
```

```python
display(dtc_df.index)
display(sensor_df.index)
```

```python
sensor_types=sensor_df["sensor_type"].value_counts(normalize=True)*100
sensor_types
```

```python
px.bar(sensor_types,color=sensor_types.index)
```

```python
agg_data=sensor_df.groupby("sensor_type")[["value"]].agg(["min","max","mean"])
display(agg_data)
agg_data.columns=["_".join(col) for col in agg_data.columns]
agg_data.columns
display(agg_data.index)
```

```python

px.bar(data_frame=agg_data,barmode="group",facet_col=agg_data.index)
```

```python
for grp in agg_data.index:
    display(px.bar(data_frame=agg_data.loc[grp],color=agg_data.columns))
```

```python
sensor_df.index=pd.to_datetime(sensor_df.index)
min_freq=sensor_df.resample("1min")[["value"]].mean()
# min_freq=min_freq.unstack(0)
min_freq
```

```python
sensor_df_agg=sensor_df.groupby("sensor_type",observed=True).resample("1min")[["value"]].mean()
sensor_df_agg=sensor_df_agg.unstack(0)
sensor_df_agg.columns=[idx[1]+"_"+idx[0] for idx in sensor_df_agg.columns]
sensor_df_agg
```

```python
for col in sensor_df_agg:
    display(px.line(sensor_df_agg,x=sensor_df_agg.index,y=sensor_df_agg[col],markers=True))
```

```python
res=dtc_df.groupby(["dtc_code","severity"])["severity"].size().unstack()
res
```

```python
merged_dff=sensor_df.merge(dtc_df,left_on="vehicle_id",right_on="vehicle_id",how="outer")
```

```python
merged_df=merged_dff.drop_duplicates()
merged_df
```

```python
grouped_df=merged_df.groupby("vehicle_id")
grouped_df
```

```python
grouped_df.groups
```

```python
excel_path="resulted_excel.xlsx"
with pd.ExcelWriter(excel_path) as excel_file:
    for grp in grouped_df.groups:
        grouped_df.get_group(grp).to_excel(excel_file,sheet_name=grp)
```

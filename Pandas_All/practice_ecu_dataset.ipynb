```python
import pandas as pd
import plotly.express as px
```

```python
ecu_df=pd.read_csv("F:\KPIT\Pandas_All\Datasets\ecu_sensor_data.csv")
ecu_df["unit"]=pd.Categorical(ecu_df["unit"])
```

```python
ecu_df.sample(5)
```

```python
ecu_df.columns
```

```python
ecu_df.index
```

```python
# Display the total number of unique vehicles and sensor types in the dataset.
unique_vehicle=ecu_df["vehicle_id"].nunique()
unique_sensor_type=ecu_df["sensor_type"].nunique()
display(unique_vehicle)
display(unique_sensor_type)
```

```python
#Find the top 3 most frequently recorded sensor types.
freq=ecu_df.groupby("sensor_type").size()
freq_df=pd.DataFrame(freq)
freq_df=freq_df.rename(columns={0:"Count"})
freq_df.sort_values("Count",ascending=False).head(3)
```

```python
# Calculate the mean, median, and standard deviation of sensor value for each sensor_type.
ecu_df.groupby("sensor_type").describe()
```

```python
# Identify which vehicle_id has the highest overall average sensor value.
mx=ecu_df.groupby("vehicle_id")["value"].max()
mx=pd.DataFrame(mx)
mx.sort_values("value",ascending=False).head(1)
```

```python
# Plot a boxplot (using Plotly Express) showing the distribution of value across all sensor_types.
res=ecu_df["sensor_type"].value_counts()
res=pd.DataFrame(res)
display(res)
px.box(data_frame=ecu_df,x="sensor_type",y="value",points="all")
```

```python
# Filter all records where the sensor_type is "coolant_temp" and the value exceeds 95°C.
display(ecu_df.sample(5))
ecu_df[(ecu_df["sensor_type"]=="coolant_temp") & (ecu_df["value"]>95) & (ecu_df["unit"] == "Celsius")]
```

```python
# Retrieve all rows where vehicle_id is "VEH002" and timestamp is between "2025-10-01 09:00:00" and "2025-10-01 09:15:00".
ecu_df["timestamp"] = pd.to_datetime(ecu_df["timestamp"])
ross=ecu_df[(ecu_df["vehicle_id"]=="VEH002") & (ecu_df["timestamp"] > "2025-10-01 09:00:00") & (ecu_df["timestamp"] < "2025-10-01 09:15:00")]
ross
```

```python
# Find all instances where sensor_type is "engine_rpm" and value is greater than 4000 RPM.
ecu_df[(ecu_df["sensor_type"] == "engine_rpm") & (ecu_df["value"] > 4000) & (ecu_df["unit"] == "RPM")]
```

```python
# Group the data by vehicle_id and sensor_type, and find the average value for each combination.
ecu_df.groupby(["vehicle_id","sensor_type"])["value"].mean()
```

```python
# Create a pivot table with vehicle_id as rows, sensor_type as columns, and mean value as the values.
pd.pivot_table(data=ecu_df,index="vehicle_id",values="value",aggfunc="mean")
```

```python
# Plot a bar graph (using Plotly Express) showing the average sensor value per sensor_type.
avg=ecu_df.groupby("sensor_type")["value"].mean()
avg=pd.DataFrame(avg)
px.bar(data_frame=avg,color=avg.index)
```

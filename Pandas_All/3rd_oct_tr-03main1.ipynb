```python
import pandas as pd
import plotly.express as px
```

```python
# Convert columns to categorical
def convert_columns_to_categorical(df:pd.DataFrame,columns:list[str])->pd.DataFrame:
    for col in columns:
        df[col]=pd.Categorical(df[col])    # Think of exceptional handling
    
    return df
```

```python
# Exploring the Database
def explore_data(df:pd.DataFrame)->None:
    print(f"Samples of the dataset")
    display(df.sample(5))
    print(f"Shape of Data is {df.shape}")
    print(f"List of columns is {df.columns}")
    print(f"Information of the dataset is {df.info()}")
    print(f"Descriptive statistics of real-value columns :")
    display(df.describe())
    print(f" Missing value count is {df.isna().sum()}")
```

```python
dtc_code_path=r"F:\KPIT\Pandas_All\Datasets\dtc_codes.csv"
sensor_data_file_path=r"F:\KPIT\Pandas_All\Datasets\ecu_sensor_data.csv"
```

```python
sensor_df=pd.read_csv(sensor_data_file_path,index_col="timestamp")
sensor_df=convert_columns_to_categorical(sensor_df,["sensor_type","unit"])
sensor_df.index=pd.to_datetime(sensor_df.index)
dtc_df=pd.read_csv(dtc_code_path,index_col="timestamp")
dtc_df=convert_columns_to_categorical(dtc_df,["severity","dtc_code"])
```

### Average minimum, maximum values for sensor reading ("value") across each category

```python
result=sensor_df.groupby("sensor_type")[["value"]].agg(["min","max","mean"])
# print(result.columns) # Multi valued index , that's why we have joined it.
# print(sensor_df.index)
# Converting the Multi-Index into the single one.
result.columns=["_".join(col) for col in result.columns]
# display(result)

mapping={
    "coolant_temp":"Celsius",
    "engine_rpm":"RPM",
    "oxygen_sensor":"Volts",
    "throttle_position":"pct(%)"
}
# px.bar(data_frame=result,barmode="group",facet_col=result.index)
print(result.index)
for grp in result.index:
    # display(px.bar(result,y=grp,color=result.index))
    display(px.bar(result.loc[grp],y=grp,color=result.columns,labels={'index':mapping[grp]}))
```

### Join Dataframes based on timestamps

```python
sensor_df.merge(dtc_df,left_on="vehicle_id",right_on="vehicle_id",how="outer")
```

```python
# The above data contains duplicates
# We use drop duplicates to drop the duplicate rows
merged_df=sensor_df.merge(dtc_df,left_on="vehicle_id",right_on="vehicle_id",how="outer").drop_duplicates()

```

### Grouped the merged data based on the vehicle ID. Expor the grouped data as a separate sheet in an excel file.

### For each State, calculate annual revenue (sum of Selling Price) and the year‑over‑year growth %

```python
# annual_revenue=df.groupby("State").agg(
#     Annual_Revenue=("Selling Price","sum"),
# )
# annual_revenue["year_over_growth"]=df[]


# Ensure Sale Date is datetime
df['Sale Date'] = pd.to_datetime(df['Sale Date'])
df['Sale_Year'] = df['Sale Date'].dt.year

# Group by State and Year
annual_revenue = df.groupby(["State", "Sale_Year"]).agg(
    Annual_Revenue=("Selling Price", "sum")
).reset_index()




# Sort by State and Year
annual_revenue = annual_revenue.sort_values(["State", "Sale_Year"])

# Calculate year-over-year growth %
annual_revenue["YoY_Growth_%"] = annual_revenue.groupby("State")["Annual_Revenue"].pct_change() * 100

annual_revenue["YoY_Growth_%"]=annual_revenue["YoY_Growth_%"].fillna(0)
annual_revenue
```

### Calculate a centered 6‑month rolling average of Odometer based on Sale Date.

```python
monthly_odo=df.resample("ME")["Odometer"].mean()
roll_df=monthly_odo.rolling(window=6,center=True).mean()
roll_df=roll_df.to_frame(name="6 month rolling average")
roll_df
```

### For each month, extract the Selling Price of the first sale (minimum Sale Date within that month).

### Generate a cumulative revenue series ordered by Sale Date.

```python
ndres=df.resample("M")["Selling Price"].sum()
cumm_revenue=ndres.cumsum()
final_res=cumm_revenue.to_frame(name="Cummulative Revenue")
final_res=final_res.reset_index()
final_res["Month"]=final_res["Sale Date"].dt.to_period("M")
final_res=final_res.drop("Sale Date",axis=1)
final_res.set_index("Cummulative Revenue")
display(final_res)


bc=df.resample("ME")["Selling Price"].sum()
cumm=bc.cumsum()
cumm=pd.DataFrame(cumm)
cumm=cumm.reset_index()
cumm["Sale Date"]=cumm["Sale Date"].dt.to_period("M")
cumm=cumm.rename(columns={"Sale Date":"Month","Selling Price":"Cummulative Revenue"})
cumm=cumm[["Cummulative Revenue","Month"]]
display(cumm)
```

```python
import pandas as pd
import plotly.express as px

# Loading the dataset
car_sales_df=pd.read_csv("/home/pragyanp/Desktop/Pandas_NLP_DL/Datasets/car_sales_usa_data 1.csv",parse_dates=["Sale Date"])
car_sales_df

```

```python
# Function to explore the dataset
def explore_data(car_sales_df:pd.DataFrame)->None:
    print(f"{car_sales_df.sample(3)}")
    print(f"Info of the dataset : {car_sales_df.info()}")
    print(f"Description of the dataset : {car_sales_df.describe()}")
    print(f"Index of the dataset : {car_sales_df.index}")
    print(f"All the columns of the dataset : {car_sales_df.columns}")

# Function to conver the column into the categorical form
def convert_categorical(car_sales_df:pd.DataFrame , columns:list[str])->pd.DataFrame:
    for col in columns:
        car_sales_df[col]=pd.Categorical(car_sales_df[col])

    return car_sales_df
```

```python
explore_data(car_sales_df)

car_sales_df=convert_categorical(car_sales_df,["Transmission","Body","Make"])
```

### Produce a quarter‑wise (3‑month) count of sales transactions across the whole dataset.

```python
new_df=car_sales_df
new_df=new_df.set_index("Sale Date")
resampled_df=new_df.resample("3ME")["Selling Price"].size()
resampled_df=pd.DataFrame(resampled_df)
resampled_df=resampled_df.rename(columns={0:"Count"})
resampled_df.index=resampled_df.index.to_period("Q")
resampled_df.index=[f"{p.year}-Q{p.quarter}" for p in resampled_df.index]
resampled_df
```

### For each State, calculate annual revenue (sum of Selling Price) and the year‑over‑year growth %. 

```python
new_df=car_sales_df
new_df["Sale Year"]=car_sales_df["Sale Date"].dt.year
new_df=car_sales_df.groupby(["State","Sale Year"])[["Selling Price"]].sum()
new_df["year_over_growth(%)"]=new_df["Selling Price"].pct_change()
new_df
```

### Resample the data to a monthly frequency and compute the mean Selling Price for each month.

```python
new_df=car_sales_df
new_df["Sale Date"]=pd.to_datetime(car_sales_df["Sale Date"])
new_df=new_df.set_index("Sale Date")
display(new_df.index)
new_df=new_df.resample("ME")[["Selling Price"]].mean()
new_df["Selling Price"].fillna(0)
```

### Resample to a weekly frequency (weeks ending on Sunday) and count the number of sales per week.

```python
new_df=car_sales_df
new_df["Sale Date"]=pd.to_datetime(car_sales_df["Sale Date"])
new_df=new_df.set_index("Sale Date")
new_df=new_df.resample("W")[["Selling Price"]].size()
new_df=pd.DataFrame(new_df)
new_df=new_df.rename(columns={0:"Count"})
new_df
```

### Calculate a centered 6‑month rolling average of Odometer based on Sale Date.

```python
new_df=car_sales_df
new_df["Sale Date"]=pd.to_datetime(car_sales_df["Sale Date"])
new_df=new_df.set_index("Sale Date")
monthly_odo=new_df.resample("ME")["Odometer"].mean()
roll_df=monthly_odo.rolling(window=6,center=True).mean().to_frame()
roll_df=roll_df["Odometer"].fillna(0)
roll_df
```

### For each month, extract the Selling Price of the first sale (minimum Sale Date within that month).

```python
new_df=car_sales_df
new_df["Sale Date"]=pd.to_datetime(car_sales_df["Sale Date"])
new_df=new_df.set_index("Sale Date")
print(type(new_df.index))
new_df=new_df.resample("ME").first()[["Selling Price"]]
new_df["Selling Price"].fillna(0)
new_df.index=new_df.index.to_period("M")
new_df
```

### Generate a cumulative revenue series ordered by Sale Date.

```python
new_df=car_sales_df
new_df["Sale Date"]=pd.to_datetime(car_sales_df["Sale Date"])
new_df=new_df.set_index("Sale Date")
before_cumm_sum=new_df.resample("ME")[["Selling Price"]].sum()
cumm_sum=before_cumm_sum.cumsum()
cumm_sum.sort_values("Selling Price")
cumm_sum.rename(columns={"Selling Price":"Cummulative Sum"},inplace=True)
cumm_sum
```

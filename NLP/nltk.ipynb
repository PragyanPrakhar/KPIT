```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import pandas as pd
import numpy as np
import regex as re
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
sample_text="Hello , My name is Pragyan Prakhar . I'm excellent at what I do."
pd.set_option('display.max_colwidth',200)
```

### Punctuation Removal

```python
import string
sample_text_without_punctuation=sample_text.translate(str.maketrans("","",string.punctuation))
print(f"String without punctuation : {sample_text_without_punctuation}")
```

### Tokenization

```python
word_tokenized_list=word_tokenize(sample_text_without_punctuation)
print(f"Word with tokenization : {word_tokenized_list}")

sent_tokenized_list=sent_tokenize(sample_text_without_punctuation)
print(f"Sent tokenized list : {sent_tokenized_list}")
```

### Stop Word Removal

```python
stop_words=set(stopwords.words('english'))
without_stop_words=[word for word in word_tokenized_list if word not in stop_words]
print(f"Without Stop Words : {without_stop_words}")
```

### Stemming

```python
ps=PorterStemmer()
stemmed_list=[ps.stem(word) for word in word_tokenized_list]
print(f"Stemmed list is {stemmed_list}")
```

### Lemmatization


```python
wnl=WordNetLemmatizer()
lemmatized_list=[wnl.lemmatize(word) for word in word_tokenized_list]
print(f"Word tokenized list is : {lemmatized_list}")
```

### Converting into the dataframe

```python
dict_nlp={
    0:{
        "Original_String":sample_text,
        "Puncutation_Removed_Text":sample_text_without_punctuation,
        "Word_Tokenized_List":word_tokenized_list,
        "Sentence_Tokenized_List":sent_tokenized_list,
        "Stemmed_List":stemmed_list,
        "lemmatized_list":lemmatized_list
    }
}
data_frame=pd.DataFrame(dict_nlp)
data_frame.index.name="Operations"
data_frame = data_frame.rename(columns={
    "index": "Operations",
    0: "Operated String"
})

display(data_frame)
display(data_frame.index)
display(data_frame.columns)

```

```python
# Extract numeric digits
import re
text = "Order 123 costs 456 rupees"
patt=re.compile(r"\d+")
print(re.findall(patt,text))
```

```python
# Match all words
text = "Regex is fun!"
patt=re.compile(r"\w+")
print(re.findall(patt,text))
```

```python
# Match all white spaces
text = "Hello   World\nRegex"
pat=re.compile(r"\W")
print(re.findall(pat,text))
```

```python
# Match words starting with upper case
text = "I like Python and Regex"
pat=re.compile(r"\b[A-Z]\w*")
print(re.findall(pat,text))
```

```python
# Extract all email addresses
text = "Contact me at abc@mail.com or xyz@domain.org"
pat=re.compile(r"[a-zA-Z0-9._%+-]+@[a-zA-Z0-9._]+\.[a-z]{2,}")
print(re.findall(pat,text))
```

```python
# Extract all hashtags
import re
text = "Loving #python and #regex"
pat=re.compile(r"#\w+")
print(re.findall(pat,text))
print(help(re))
```

```python
#Extract all numbers with decimal
text = "The values are 3.14, 2.71, and 42"
pat=re.compile(r"\d+\.\d+")
print(re.findall(pat,text))
```

```python
#Extract valid phone numbers (+91-xxxxxxxxxx)
text = "Call me at +91-9876543210 or +91-1234567890"
pat=re.compile(r"\+91-\d{10}")
print(re.findall(pat,text))
```

```python
# Extracrting all the word that ends with ing
text = "I am playing, running, and eating"
patt=re.compile(r"\b\w+ing\b")
print(re.findall(patt,text))
```

```python
#Extract dates in DD/MM/YYYY
text = "Today is 10/09/2025 and tomorrow 11/09/2025"
pat=re.compile(r"\d{2}/\d{2}/\d{4}")
print(re.findall(pat,text))
```

```python
#Match "color" or "colour"
text = "My favorite color is red and favourite colour is blue"
pat=re.compile(r"colour|color")
print(re.findall(pat,text))
```

```python
#Extract text inside double quotes
text = 'She said "hello" and then "bye"'
pat=re.compile(r'"(\w+)"')  
print(re.findall(pat,text))
```

```python
#Extracting all capital words
text = "My name is PRAGYAN PRAKHAR"
pattern = r"\b[A-Z]{2,}\b"
print(re.findall(pattern,text))
```

```python
#Extracting hash color nodes
text = "Colors are #FF5733, #C70039, and #900C3F"
pat=re.compile(r"\#[A-Z0-9]+")
patt=re.compile(r"#\w+")

print(re.findall(pat,text))
```

```python
# Extracting file extensions
text = "file1.txt, file2.pdf, image.jpeg"
pat=re.compile(r"\.[a-z]+")
print(re.findall(pat,text))
```

```python
# Extracting first name and last name
text = "Name: Pragyan Prakhar"
pat=re.compile(r"Name: (\w+)\s(\w+)")
print(re.findall(pat,text))
```
